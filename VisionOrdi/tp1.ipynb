```python
import matplotlib.pyplot as plt
def main():
    # Initialize a new Plotting window
    plt.figure(figsize=(10, 10))
    # Initializing 3D capabilities
    axes = plt.axes(projection="3d", proj_type='ortho')
    # Setting axis properties
    axes.set_xlim(-10, 10) # X Axis graduation
    axes.set_ylim(-10, 10) # Y Axis graduation
    axes.set_zlim(-10, 10) # Z Axis graduation
    axes.set_xlabel('X') # X Axis label
    axes.set_ylabel('Y') # Y Axis label
    axes.set_zlabel('Z') # Z Axis label
    axes.xaxis.label.set_color('red') # X Axis color
    axes.yaxis.label.set_color('green') # Y Axis color
    axes.zaxis.label.set_color('blue') # Z Axis color
    axes.tick_params(axis='x', colors='red') # X Axis graduation color
    axes.tick_params(axis='y', colors='green') # Y Axis graduation color
    axes.tick_params(axis='z', colors='blue') # Z Axis graduation color
    # Display the 3D plotting window
    plt.show()
if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))
axes = plt.axes(projection="3d", proj_type='ortho')
axes.set_xlim(-10, 10) 
axes.set_ylim(-10, 10)
axes.set_zlim(-10, 10)
axes.set_xlabel('X') 
axes.set_ylabel('Y')
axes.set_zlabel('Z')
axes.xaxis.label.set_color('red')
axes.yaxis.label.set_color('green') 
axes.zaxis.label.set_color('blue')
axes.tick_params(axis='x', colors='red') 
axes.tick_params(axis='y', colors='green') 
axes.tick_params(axis='z', colors='blue') 

lim = 10

plt.plot([0,lim],[0,0],[0,0], color='red', linestyle='solid')
plt.plot([0,-lim],[0,0],[0,0], color='red', linestyle='dashed')

plt.plot([0,0],[0,lim],[0,0], color='green', linestyle='solid')
plt.plot([0,0],[0,-lim],[0,0], color='green', linestyle='dashed')

plt.plot([0,0],[0,0],[0,lim], color='blue', linestyle='solid')
plt.plot([0,0],[0,0],[0,-lim], color='blue', linestyle='dashed')

import math

def rot_x_point(point, omega):
    x = point[0]
    y = point[1]
    z = point[2]
    
    xr = x
    yr = y * math.cos(omega) - z * math.sin(omega)
    zr = y * math.sin(omega) + z * math.cos(omega)
    return (xr,yr,zr)

def rot_y_point(point, phi):
    x = point[0]
    y = point[1]
    z = point[2]
    
    xr = x * math.cos(phi) + z * math.sin(phi)
    yr = y 
    zr = z * math.cos(phi) - x * math.sin(phi)
    return (xr,yr,zr)

def rot_z_point(point, kappa):
    x = point[0]
    y = point[1]
    z = point[2]
    
    xr = x * math.cos(kappa) - y * math.sin(kappa)
    yr = x * math.sin(kappa) + y * math.cos(kappa)
    zr = z
    return (xr,yr,zr)

point = (4.0,4.0,4.0)
rot_x = rot_x_point(point, math.pi/4)
rot_y = rot_y_point(point, math.pi/4)
rot_z = rot_z_point(point, math.pi/4)

plt.plot(point[0], point[1], point[2], marker='o', color='black')
plt.plot(rot_x[0], rot_x[1], rot_x[2], marker='o', color='red')
plt.plot(rot_y[0], rot_y[1], rot_y[2], marker='o', color='green')
plt.plot(rot_z[0], rot_z[1], rot_z[2], marker='o', color='blue')



def rot_point(point, omega, phi, kappa):
    x = point[0]
    y = point[1]
    z = point[2]
    
    cos_phi = math.cos(phi)
    cos_kappa = math.cos(kappa)
    cos_omega = math.cos(omega)
    
    sin_phi = math.sin(phi)
    sin_kappa = math.sin(kappa)
    sin_omega = math.sin(omega)
    
    val1 = cos_phi*cos_kappa
    val2 = (sin_omega*sin_phi*cos_kappa - cos_omega*sin_kappa)
    val3 = (sin_omega*sin_phi*sin_kappa + cos_omega*cos_kappa)
    val4 = (cos_omega*sin_phi*cos_kappa + sin_omega*sin_kappa)
    val5 = (cos_omega*sin_phi*sin_kappa - sin_omega*cos_kappa)

    xr = x * val1 + y * val2 + z * val4
    yr = x * cos_phi*sin_kappa + y * val3 + z * val5
    zr = -x * sin_phi + y * sin_omega*cos_phi + z * cos_omega*cos_phi
    
    return (xr, yr, zr)

# rotation_point = rot_point(point, math.pi/4,math.pi/4,math.pi/4)
# rotation_point = rot_point(point, math.pi/4, 0, 0)
rotation_point = rot_point(point, 0, math.pi/4, 0)
# rotation_point = rot_point(point, 0, 0, math.pi/4)
print(rotation_point)
plt.plot(rotation_point[0], rotation_point[1], rotation_point[2], marker='x', color='orange')

def rot_point_comp(point, omega, phi, kappa):
    return rot_z_point(rot_y_point(rot_x_point(point, omega),phi),kappa)

rotation_compo = rot_point_comp(point, 0, math.pi/4, 0)
print(rotation_compo)
plt.plot(rotation_compo[0], rotation_compo[1], rotation_compo[2], marker='o', color='grey')

plt.show()

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))
axes = plt.axes(projection="3d", proj_type='ortho')
axes.set_xlim(-10, 10) 
axes.set_ylim(-10, 10)
axes.set_zlim(-10, 10)
axes.set_xlabel('X') 
axes.set_ylabel('Y')
axes.set_zlabel('Z')
axes.xaxis.label.set_color('red')
axes.yaxis.label.set_color('green') 
axes.zaxis.label.set_color('blue')
axes.tick_params(axis='x', colors='red') 
axes.tick_params(axis='y', colors='green') 
axes.tick_params(axis='z', colors='blue') 
import math

def cube(s):
    return [
        (-s, -s, -s),
        (s, -s, -s),
        (-s, s, -s),
        (s, s, -s),
        (-s, -s, s),
        (s, -s, s),
        (-s, s, s),
        (s, s, s),
    ]

def edge(pt1, pt2):
    return [pt1[0], pt2[0]], [pt1[1], pt2[1]], [pt1[2], pt2[2]]

def display_cube(vertices):
    x_edges = [(0, 1), (3, 2), (4, 5), (7, 6)]
    for i, j in x_edges:
        plt.plot(*edge(vertices[i], vertices[j]), color="red")

    y_edges = [(0, 2), (1, 3), (4, 6), (5, 7)]
    for i, j in y_edges:
        plt.plot(*edge(vertices[i], vertices[j]), color="green")

    z_edges = [(0, 4), (1, 5), (2, 6), (3, 7)]
    for i, j in z_edges:
        plt.plot(*edge(vertices[i], vertices[j]), color="blue")
        

def translate_cube(vertices, alpha, beta, gamma):
    res = []
    vecteur = (alpha, beta, gamma)
    for i in vertices :
        res+= translate_point(i, vecteur)
    return res

def rotate_cube(vertices, omega, phi, kappa):
    res = []
    for i in vertices:
        res += [rot_point_comp(i, omega, phi, kappa)]
    return res

vertices = cube(3.0)
# alpha = 1.0
# beta = 2.0
# gamma = 3.0
# translated_vertices = translate_cube(vertices, alpha, beta, gamma)
        
# display_cube(vertices)
# display_cube(translated_vertices)

translated_vertices = translate_cube(vertices, 1, 2, 3)
# display_cube(translated_vertices)

rot_vertices = rotate_cube(vertices, math.pi / 4, math.pi / 3, math.pi / 2)
display_cube(rot_vertices)

cubee = cube(2)
display_cube(cubee)
alpha, beta, gamma = 0.25, 0.50, 0.75
omega = math.pi / 6
phi = math.pi / 4
kappa = math.pi / 3

tcube = translate_cube(cubee, alpha, beta, gamma)
rcube = rotate_cube(tcube, omega, phi, kappa)
display_cube(rcube)

rcube1 = rotate_cube(cubee, omega, phi, kappa)
tcube1 = translate_cube(rcube1, alpha, beta, gamma)
display_cube(tcube1)



plt.show()

```
